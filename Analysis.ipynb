```python
## Stata-style frequency table: Frequency, Percent, Cumulative Percent and Total.

def tab(variable, data, missing=False):

    x = data[variable]                      # Select the variable

    # Count frequencies
    freq = x.value_counts(
        dropna=not missing,
        sort=False
    )

    # Calculate percentages
    percent = freq / freq.sum() * 100

    # Calculate cumulative percentages
    cumulative = percent.cumsum()

    # Create table
    table = pd.DataFrame({
        "Frequency": freq,
        "Percent (%)": percent.round(2),
        "Cumulative (%)": cumulative.round(2)
    })

    # Add total row
    table.loc["Total"] = [
        freq.sum(),
        round(percent.sum(), 2),
        round(cumulative.iloc[-1], 2)
    ] if len(freq) else [0, 0, 0]

    # Display results
    print(f"\nFrequency distribution: {variable}")
    display(table)
```
